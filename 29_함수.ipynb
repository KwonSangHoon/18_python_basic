```python
# 덧셈함수 만들기

# def 함수이름(매개변수, 매개변수2):
#     매개변수를 사용한 코드

def add(a, b):
    # a = 1, b = 20
    print(a + b)

result1 = add(10, 20)
print(result1)
```

```python
# 함수의 결과를 반환하기
# def 함수이름(매개변수, 매개변수2):
#     return 매개변수를 사용한 코드

def add(a, b):
    return a + b

result2 = add(10, 20)
print(result2)
```

```python
# return 으로 함수 중간에서 빠져나오기
def not_ten(a):
    if a == 10:
        return 
    print(a, "입니다", sep="")
```

```python
# 함수의 호출 과정 알아보기
def mul(a, b):
    c = a * b
    return c

def add(a, b):
    c = a + b
    print(c)
    d = mul(a, b)
    print(d)

x = 10
y = 20
add(x, y)
```

```python
# 연습문제: 몫과 나머지를 구하는 함수 만들기

x = 10
y = 3

def get_quotient_remainder(a, b):
    return a // b, a % b

quotient, remainder = get_quotient_remainder(x, y)
print('몫: {0}, 나머지: {1}' .format(quotient, remainder))
```

```python
# 심사문제: 사칙 연산 함수 만들기

x, y = map(int, input().split())

def calc(x, y):
    return x + y, x - y, x * y, x / y

a, s, m, d = calc(x, y)
print('덧셈: {0}, 뺄셈: {1}, 곱셈: {2}, 나눗셈: {3}'.format(a, s, m, d))
```

```python
x, y = map(int, input().split())

def calc(num_a, num_b):
    add = num_a + num_b
    sub = num_a - num_b
    mul = num_a * num_b
    div = num_a / num_b
    return add, sub, mul, div

a, s, m, d = calc(x, y)
print("덧셈: {0}, 뺄셈: {1}, 곱셈: {2}, 나눗셈: {3}".format(a, s, m, d))
```
