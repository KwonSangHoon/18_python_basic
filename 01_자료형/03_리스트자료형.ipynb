```python
## 리스트 관련 함수

## 리스트에 요소 추가하기 - append
a = [1, 2, 3]
a.append(4)
print(a)
```

```python
a.append(["a", "b"])
print(a)
```

```python
## 리스트정렬 - sort
a = [1, 4, 3, 2]
a.sort()
a
```

```python
a = ["a", "c", "b"]
a.sort(reverse=True)
a
```

```python
## 리스트 뒤집기 - reverse
a = ["a", "c", "b"]
a.reverse()
a
```

```python
# 인덱스 반환 - index
a = [1, 2, 3, 4]
a.index(3)
```

```python
## 리스트에 요소 삽입 - insert
a = [1, 2, 3]
a.insert(0, 4)
a.insert(4, 5)
a
```

```python
## 리스트에 요소제거 - remove()
a = [1, 2, 3, 1, 2 ,3]
a.remove(2)
a.remove(2)
a
```

```python
## 리스트 요소 꺼내기 - pop
a = [1, 2, 3]
number = a.pop()

print(number)
print(a)
```

```python
## 리스트에 포함된 요소 x의 개수 세기 - count
a = [1, 2, 3, 4, 3]

a.count(3)
```
