```python
# 집합 연산 사용하기
## 합집합
## 세트1 | 세트2
## set.union(세트1, 세트2) = 세트1.union(세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)
print(set.union(a, b))
print(a.union(b))
```

```python
# 교집합
## 세트1 & 세트2
## set.intersection(a, b)
## a.intersection(b)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a & b)
print(set.intersection(a, b))
print(a.intersection(b))
```

```python
# 차집합
## a - b
## set.difference(a, b)
## a.difference(b)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a - b)
print(set.difference(a, b))
print(a.difference(b))
print("="*20)
print(b - a)
```

```python
# 대칭차집합
## 세트1 ^ 세트2
## set.symmetric_difference(a, b)
print(a ^ b)
print(set.symmetric_difference(a, b))

print(a)
```

```python
# 집합 연산 후 할당 연산자 사용하기
## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}
a |= {5}
print(a)

a = {1, 2, 3, 4}
a.update ({5})
print(a)
```

```python
# intersection_update
## 세트1 &= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}
print(a)

a = {1, 2, 3, 4}
a.intersection_update({0, 1, 2, 3, 4})
print(a)
```

```python
a = {1, 2, 3, 4}
a ^= {3, 4, 5, 6}
print(a)
```

```python
# 연습문제: 공배수 구하기

a = {i for i in range(1, 101) if i % 3 == 0}
b = {i for i in range(1, 101) if i % 5 == 0}

print(a & b)
```

```python
# 심사문제: 공약수 구하기

a, b = map(int, input().split())

divisor_a = {i for i in range(1, a + 1) if a % i == 0}
divisor_b = {i for i in range(1, b + 1) if b % i == 0}
divisor = divisor_a & divisor_b

result = 0
if type(divisor) == set:
    result = sum(divisor)

print(result)
```
