```python
# 딕셔너리 표현식 사용하기
# {키:값 for 키, 값 in 딕셔너리}

keys = ["a", "b", "c", "d"]
imsi = dict.fromkeys(keys)
imsi

x ={key:value for key, value in imsi.items()}
x
```

```python
# 딕셔너리 표현식에서 if 조건문 사용하기
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

for key, value in x.items():
    if value == 20:
        del x[key]

print(x)
```

```python
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

del x["b"]
print(x)
```

```python
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}
y = x.copy()

for key, value in y.items():
    if value == 20:
        del x[key]

print(x)
```

```python
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

x = {key: value for key, value in x.items() if value != 20}
```

```python
# 심사문제: 딕셔너리에서 특정 값 삭제하기

keys = input().split()
values = map(int, input().split())

x = dict(zip(keys, values))

x.pop('delta', None)

x = {key: value for key, value in x.items() if value != 30}

print(x)
```

```python
# 심사문제
string = input().split()
num = map(int, input().split())
x = dict(zip(string,num))

result = {key:value for key,value in x.items() if value != 30 and key != "delta"}
print(result)
```
