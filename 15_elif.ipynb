```python
# 음료수 자판기
button = int(input())

if button == 1:
    print("콜라")
elif button == 2:
    print("사이다")
elif button == 3:
    print("환타")
else :
    print("제공하지 않는 메뉴")
```

```python
# 연습문제 : if, elif, else 모두 사용하기

x = int(input())

if 11 <= x <= 20:
    print('11~20')
elif 21 <= x <= 30:
    print('21~30')
else:
    print('아무것도 해당하지 않음')
```

```python
# 심사문제: 교통카드 시스템 만들기

age = int(input())

money = 9000

if age >= 7 and age <= 12:
    money -= 650
elif age >= 13 and age <= 18:
    money -= 1050 
elif age >= 19:
    money -= 1250

print(money)
```
