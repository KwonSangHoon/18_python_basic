```python
# 재귀호출에 종료 조건 만들기
def hello(count):
    if count == 0:
        return

    print("Hello world", count)

    count -= 1
    hello(count)
```

```python
hello(5)
```

```python
# 재귀호출로 팩토리얼 구하기

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)
```

```python
# 5 : 5 * factorial(4)
# 4 : 5 * 4 * factorial(3)
# 3 : 5 * 4 * 3 * factorial(2)
# 2 : 5 * 4 * 3 * 2 * factorial(1)
# 1 : 5 * 4 * 3 * 2 * 1
print(factorial(5))
```

```python
# 연습문제: 재귀호출로 회문 판별하기

def is_palindrome(word):

    if len(word) < 2:
        return True
    if word[0] != word[-1]:
        return False
    return is_palindrome(word[1:-1])

print(is_palindrome("hello"))
print(is_palindrome("level"))
```

```python
# 심사문제: 재귀호출로 피보나치 수 구하기

def fib(n):
    a, b = 0, 1
    for i in range(n):
        a, b = b, a + b
    return a

n = int(input())
print(fib(n))
```

```python
def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)

n =  int(input())
print(fib(n))
```
