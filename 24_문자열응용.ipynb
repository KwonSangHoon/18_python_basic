```python
# format 매서도 사용하기
print("Hello, {0}".format("world"))
```

```python
# format 매서드로 값을 여러개 넣기
print("Hello {0} {2} {1}".format("python", "Script", 3.6))
```

```python
# format 매서드로 같은 값을 여러개 넣기
print("{0} {0} {1} {1}".format("Python", "Script"))
```

```python
# format 매서드에 인덱스 생략하기
print("Hello {} {} {}".format("Python", "Script", 3.6))
```

```python
# format 매서드에서 인덱스 대신 이름 지정하기
print("Hello {language} {version}".format(language="Python", version=3.6))
```

```python
# 문자열 포매팅에 변수를 그대로 사용하기
language = "Python"
version = 3.6

print(f"Hello {language} {version}")
```

```python
# format 매서드로 문자열 정렬하기
# "{0:<10}".format("python")
print("{0:<10}".format("python"))
print("{0:>10}".format("python"))
print("{:>0}".format("python"))
```

```python
# 숫자 개수 맞추기
# "%0개수d"
# "{인덱스:0개수d}".format(숫자):
print("%03d" % 1)
print("{:03d}".format(35))
```

```python
# 심사문제 1: 특정 단어 개수 세기

import re

text = input()

words = re.findall(r'\bthe\b', text)

print(len(words))
```

```python
# 심사문제 2: 높은 가격순으로 출력하기

prices = input().split(';')

prices = list(map(int, prices))
prices.sort(reverse=True)

for price in prices:
    print(f'{price:>9,}')
```

```python
# 심사문제-1

string = input()

chk = string.split()

print(chk.count("the"))
```

```python
# 심사문제-2

money = map(int,input().split(";"))

money = sorted(money,reverse=True)
for i in money:
    print("{:,d}".format(i))
```

```python
# 심사문제-2

a = input().split()

count = 0

for i in a:
    a = i.strip(",.'-")

    if a == "the":
        count = count + 1

print(count)
```

```python
# 심사문제-2

a = [i.strip(",.'-") for i in input().split()]

print(a.count("the"))
```

```python
# 심사문제-1

string = input()

string = string.replace("."," ")
string = string.replace(","," ")
string = string.replace("'"," ")

chk = string.split()

print(chk.count("the"))
```
