```python
# while 반복문을 두 번 사용하기

a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end=" ")
        j += 1
    print()
    i += 1
```

```python
# 심사문제: 지뢰찾기

col, row = map(int, input().split())

matrix = []
for _ in range(row):
    matrix.append(list(input()))

for i in range(row):
    for j in range(col):
        if matrix[i][j] == '.':
            count = 0

            for di in range(-1, 2):
                for dj in range(-1, 2):
                    ni = i + di
                    nj = j + dj

                    if 0 <= ni < row and 0 <= nj < col:
                        if matrix[ni][nj] == '*':
                            count += 1

            matrix[i][j] = str(count)

for line in matrix:
    print(''.join(line))
```
