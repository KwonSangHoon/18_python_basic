```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]
for x, y in a:
    print(x, y, end=" ")
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]
for i in a:
    for j in i:
        print(j, end=" ")
    print()
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=" ")
    print()
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# while 반복문을 두 번 사용하기

a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end=" ")
        j += 1
    print()
    i += 1
```

```python
col, row = map(int, input().split())

mine = []
for _ in range(row):
    mine.append(list(input()))

for i in range(row):
    for j in range(col):
        if mine[i][j] == '.':
            count = 0

            for di in range(-1, 2):
                for dj in range(-1, 2):
                    ni = i + di
                    nj = j + dj

                    if 0 <= ni < row and 0 <= nj < col:
                        if mine[ni][nj] == '*':
                            count += 1

            mine[i][j] = str(count)

for line in mine:
    print(''.join(line))
```
