```python
# map을 사용하여 정수로 변환하기
a, b = map(int, ["10", "20"])
print(a)
print(b)
print("=====")
print(type(a))
print(type(b))
```

```python
a, b = map(int, input("숫자 두 개를 입력하세요").split())
print(a + b)
```

```python
# 입력받은 값을 콤마를 기준으로 분리하기
"10,20".split()
a, b = map(int, input("숫자 두 개를 입력하세요").split(","))
print(a + b)
```

```python
a, b, c = -10, 20, 30
a, b, c = map(int, input("숫자 세 개를 입력하세요").split())
print(a + b + c)
```

```python
# map을 사용하여 정수로 변환하기
a, b, c = map(int, ["50", "100", "None"])
print(a)
print(b)
print(c)
```
