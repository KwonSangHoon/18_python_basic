```python
# if 조건문의 동작 방식 알아보기
if True:
    print("참")
else:
    print("거짓")

if False:
    print("참")
else:
    print("거짓")

if None:
    print("참")
else:
    print("거짓")
```

```python
# 조건문에 숫자 지정하기

if 0:
    print("참")
else:
    print("거짓")

if 1:
    print("참")
else:
    print("거짓")


if 0x1f:
    print("참")
else:
    print("거짓")

if 0b1000:
    print("참")
else:
    print("거짓")

if 13.5:
    print("참")
else:
    print("거짓")
```

```python
# 연습문제

written_test = 75
coding_test = True

if written_test >= 80 and coding_test == True:
    print('합격')
else:
    print('불합격')
```

```python
# 심사문제

kor = int(input())
eng = int(input())
math = int(input())
science = int(input())

if (kor < 0) or (kor > 100) or (eng < 0) or (eng > 100) or (math < 0) or (math > 100) or (science < 0) or (science > 100):
    print("잘못된점수")
else:
    average = (kor + eng + math + science) / 4

    if average >= 80:
        print("합격")
    else:
        print("불합격")

```

```python
국어, 영어, 수학, 과학 = list(map(int, input().split()))

if 0 <= 국어 <= 100 and 0 <= 영어 <= 100 and 0 <= 수학 <= 100 and 0 <= 과학 <= 100:
    avg = (국어 + 영어 + 수학 + 과학) / 4
    if avg >= 80:
        print("합격")
    else:
        print("불합격")
else:
    print("잘못된 점수")
```
