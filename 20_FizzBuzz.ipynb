```python
# 1부터 100까지 숫자 출력하기
for i in range(1, 10+1):
    print(i)
```

```python
# 3의 배수일때 Fizz, 5의 배수일때 Buzz
for i in range(1, 20+1):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 20+1):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    elif i % 3 == 0 and i % 5 == 0: # 1 % 15 == 0
        print("FizzBuzz")
    else:
        print(i)
```

```python
for i in range(1, 20+1):
    if i % 3 == 0 and i % 5 == 0: # 1 % 15 == 0
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 코드 단축하기
for i in range(1, 30+1):
    print("Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or 1)
```

```python
# 연습문제: 2와 11의 배수, 공배수 처리하기

for i in range(1, 101):
    if i % 2 == 0 and i % 11 == 0:
        print("FizzBuzz")
    elif i % 2 == 0:
        print("Fizz")
    elif 1 % 11 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 심사문제: 5와 7의 배수, 공배수 처리하기

a, b = map(int, input().split())

for i in range(a, b + 1):
    if i % 5 == 0 and i % 7 == 0:
        print("FizzBuzz")
    elif i % 5 == 0:
        print("Fizz")
    elif i % 7 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 심사문제: 5와 7의 배수, 공배수 처리하기

first, second = map(int, input().split())

for i in range(first, second + 1):
    if i % 35 == 0:
        print("FizzBuzz", end=" ")
    elif i % 5 == 0:
        print("Fizz", end=" ")
    elif i % 7 == 0:
        print("Buzz", end=" ")
    else:
        print(i, end=" ")
```
