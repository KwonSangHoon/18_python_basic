```python
# filter
# filter(함수, 반복가능한객체)

def f(x):
    return x > 5 and x < 10

a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]

f(8)
list(filter(f, a))
```

```python
a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]
filter(lambda x: x > 5 and x < 10, a)
```

```python
# reduce 사용하기
# import random
# random.randint(1, 10)
# from random import randint
# randint(1, 10)

from functools import reduce
# reduce(함수, 반복가능한객체)

def f(x, y):
    return x + y

a = [1, 2, 3, 4, 5]
reduce(f, a)
```

```python
a = [1, 2, 3, 4, 5]
from functools import reduce

reduce(lambda x, y: x + y, a)
```

```python
# 연습문제: 이미지 파일만 가져오기

files = ["font", "1.png", "10.jpg", "11.gif", "2.jpg", "3.png", "table.xslx", "spec.docx"]
print(list(filter(lambda x: x.find(".jpg") != -1 or x.find(".png") != -1, files)))
```

```python
# 심사문제: 파일 이름을 한꺼번에 바꾸기

files = input().split()
print(list(map(lambda x: f"{x.split(".")[0]:0>3}.{x.split(".")[1]}", files)))
```
